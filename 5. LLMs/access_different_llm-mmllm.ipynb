Hugging-Face

Text to Text

```python
import os
HF_TOKEN = os.environ.get("HF_TOKEN")
if HF_TOKEN:
    os.environ["HF_TOKEN"] = HF_TOKEN
```

```python
from langchain_huggingface import HuggingFaceEndpoint
```

```python
llm = HuggingFaceEndpoint(
    repo_id="deepseek-ai/DeepSeek-V4-Pro",
    task="text-generation",
    max_new_tokens=512
)
```

```python
from langchain_huggingface import ChatHuggingFace

chat_model = ChatHuggingFace(llm=llm)
```

```python
from langchain_core.messages import SystemMessage, HumanMessage
messages = [
    SystemMessage(content="You're a helpful assistant"),
    HumanMessage(content="What happens when an unstoppable force meets an immovable object?")
]
```

```python
ai_msg = chat_model.invoke(messages)
```

```python
print(ai_msg.content)
```

Image to text

```python
import base64
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.messages import HumanMessage

# 1. Image ko base64 me convert karo
def encode_image(image_path):
    with open(image_path, "rb") as image_file:
        return base64.b64encode(image_file.read()).decode("utf-8")

base64_image = encode_image(r"C:\Users\yashr\OneDrive\Desktop\ai fs\5. LLMs\dogs.png")
```

```python
# 2. Gemini model initialize karo
llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    google_api_key=GOOGLE_API_KEY
)
```

```python
# 3. Text + image message banao
message = HumanMessage(
    content=[
        {
            "type": "text",
            "text": "Describe this image in detail."
        },
        {
            "type": "image_url",
            "image_url": f"data:image/png;base64,{base64_image}"
        }
    ]
)
```

```python
# 4. Model invoke karo
response = llm.invoke([message])
```

```python
print(response.content)
```

Audio → Text

```python
import base64
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.messages import HumanMessage
```

```python
def encode_audio(audio_path):
    with open(audio_path, "rb") as audio_file:
        return base64.b64encode(audio_file.read()).decode("utf-8")
```

```python
base64_audio = encode_audio(r"C:\Users\yashr\OneDrive\Desktop\ai fs\5. LLMs\audio.mp3")
```

```python
llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    google_api_key=GOOGLE_API_KEY
)
```

```python
message = HumanMessage(
    content=[
        {
            "type": "text",
            "text": "Transcribe this audio into English text."
        },
        {
            "type": "media",
            "mime_type": "audio/mp3",
            "data": base64_audio
        }
    ]
)
```

```python
response = llm.invoke([message])
```

```python
print(response.content)
```
