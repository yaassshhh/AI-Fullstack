```python
folder_path = r"C:\Users\yashr\OneDrive\Desktop\ai fs\2. Word2Vec Embedding\data"
story = []
for filename in os.listdir(folder_path):
  file_path = os.path.join(folder_path,filename)
  with open(file_path,encoding='unicode_escape') as f:
    corpus = f.read()
  raw_sent = sent_tokenize(corpus)
  for sent in raw_sent:
    story.append(simple_preprocess(sent))
```

```python
story
```

```python
len(story)
```

# Word2Vec(..., sg=0)  # CBOW #by default
# Word2Vec(..., sg=1)  # Skip-gram

```python
custom_model = gensim.models.Word2Vec(
    window=10,
    min_count=5,
    vector_size=100
)
## these all the value are the hyperparameter
```

```python
custom_model.build_vocab(story)
```

```python
custom_model.corpus_count
```

```python
custom_model.train(story,total_examples=custom_model.corpus_count,epochs=custom_model.epochs)

# (119, 2610) this is the shape of the word vector which we have created using our custom data and the vector size is 100 which we have mentioned in the hyperparameter and the total word in our vocabulary(unique words) is 2610 and the total word vector (sentences) which we have created is 119.
```
